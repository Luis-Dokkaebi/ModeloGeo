```python
import sys
import subprocess
import os
import pandas as pd
import geopandas as gpd
import ipywidgets as widgets
from ipyleaflet import Map, Marker, MarkerCluster, DrawControl, basemaps, AwesomeIcon
from shapely.geometry import shape, Point
from IPython.display import display
import numpy as np
import re
import hashlib
import asyncio
import nest_asyncio

nest_asyncio.apply()

try:
    from google.colab import drive
    drive.mount('/content/drive')
    from google.colab import output
    output.enable_custom_widget_manager()
except ImportError:
    pass

def install_deps():

    deps = ['langgraph', 'langchain-groq', 'langchain-core', 'ipyleaflet', 'geopandas', 'pandas', 'shapely', 'crawl4ai', 'nest_asyncio', 'playwright']
    for dep in deps:
        try:
            __import__(dep.replace('-', '_'))
        except ImportError:
            print(f"Instalando dependencia faltante: {dep}...")
            subprocess.check_call([sys.executable, "-m", "pip", "install", dep, "-q"])

    try:
        subprocess.check_call([sys.executable, "-m", "playwright", "install", "chromium"])
    except Exception as e:
        print(f"Nota: No se pudieron instalar los binarios de Playwright o ya están instalados. {e}")

install_deps()

from crawl4ai import AsyncWebCrawler, BrowserConfig, CrawlerRunConfig
from typing import TypedDict
from langgraph.graph import StateGraph, END
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_groq import ChatGroq

file_path = "/content/drive/MyDrive/ST_B2C/denue_inegi_09_.csv"
out_mensajes = widgets.Output()

try:
    with out_mensajes:
        print(f"Intentando cargar datos del DENUE desde:\n{file_path}")
    denue_inegi_09 = pd.read_csv(file_path, encoding="latin-1", low_memory=False)

    denue_inegi_09 = denue_inegi_09.dropna(subset=['latitud', 'longitud'])
    denue_inegi_09 = denue_inegi_09[(denue_inegi_09['latitud'] != 0) & (denue_inegi_09['longitud'] != 0)]

    with out_mensajes:
        print("Generando geometrías...")
    geometry = [Point(xy) for xy in zip(denue_inegi_09.longitud, denue_inegi_09.latitud)]
    gdf = gpd.GeoDataFrame(denue_inegi_09, geometry=geometry, crs="EPSG:4326")
    current_gdf = gdf.copy()

    with out_mensajes:
        out_mensajes.clear_output()
        print(" Datos cargados correctamente.")

except Exception as e:
    with out_mensajes:
        out_mensajes.clear_output()
        print(f" Error al cargar el archivo de datos: {e}")
        print("Asegúrate de que la ruta sea correcta y el archivo esté disponible.")
    gdf = gpd.GeoDataFrame(columns=['latitud', 'longitud', 'municipio', 'nombre_act', 'codigo_act', 'nom_estab', 'per_ocu', 'telefono', 'correoelec', 'www', 'geometry'])
    current_gdf = gdf.copy()

alcaldias = ['Todas'] + sorted(gdf['municipio'].dropna().unique().tolist()) if not gdf.empty else ['Todas']

dropdown_alcaldia = widgets.Dropdown(options=alcaldias, value='Todas', description='Alcaldía:', style={'description_width': 'initial'})

"""giros_construccion_y_relacionados = np.array([
    'Edificación de inmuebles comerciales y de servicios, excepto la supervisión',
    'Edificación de vivienda unifamiliar',
    'Edificación de naves y plantas industriales, excepto la supervisión',
    'Edificación de vivienda multifamiliar',
    'Construcción de obras de urbanización',
    'Construcción de carreteras, puentes y similares',
    'Trabajos de albañilería',
    'Trabajos de pintura y otros cubrimientos de paredes',
    'Minería de arena y grava para la construcción',
    'Fabricación de productos de madera para la construcción',
    'Fabricación de cemento y productos a base de cemento en plantas integradas',
    'Comercio al por mayor de cemento, tabique y grava',
    'Comercio al por menor en ferreterías y tlapalerías',
], dtype=object)"""

select_actividad = widgets.SelectMultiple(options=giros_construccion_y_relacionados.tolist(), value=[], description='Actividad(es):', style={'description_width': 'initial'})
btn_filtrar = widgets.Button(description='Aplicar Filtros', button_style='info', icon='filter')
out_resultados_poligono = widgets.Output()

cdmx_center = (19.4326, -99.1332)
m = Map(basemap=basemaps.CartoDB.Positron, center=cdmx_center, zoom=11, layout={'height': '600px', 'width': '100%'})

draw_control = DrawControl(polygon={"shapeOptions": {"color": "#2563eb", "fillOpacity": 0.2}}, polyline={}, circlemarker={}, rectangle={"shapeOptions": {"color": "#2563eb", "fillOpacity": 0.2}})
m.add_control(draw_control)

marker_cluster = MarkerCluster(markers=[])
m.add_layer(marker_cluster)

def get_color(actividad):
    colors = ['red', 'blue', 'green', 'purple', 'orange', 'darkred', 'lightred', 'beige', 'darkblue', 'darkgreen', 'cadetblue', 'darkpurple', 'pink', 'lightblue', 'lightgreen', 'gray', 'black', 'lightgray']
    if not isinstance(actividad, str): return 'blue'
    idx = int(hashlib.sha256(actividad.encode('utf-8')).hexdigest(), 16) % len(colors)
    return colors[idx]

def clean_val(val):
    if pd.isna(val) or val is None or str(val).strip() == "" or str(val).lower() == "nan":
        return "No disponible"
    return str(val)

selected_business = {}

def create_marker_click_handler(row_dict):
    def on_click(**kwargs):
        global selected_business
        selected_business = row_dict
        html_selected.value = f"""
        <b>Negocio:</b> {clean_val(row_dict.get('nom_estab'))}<br>
        <b>Tel:</b> {clean_val(row_dict.get('telefono'))}<br>
        <b>Email:</b> {clean_val(row_dict.get('correoelec'))}<br>
        <b>Web:</b> {clean_val(row_dict.get('www'))}
        """
    return on_click

def actualizar_mapa(b):
    global current_gdf
    with out_mensajes:
        out_mensajes.clear_output()
        print("Aplicando filtros y actualizando mapa...")
    alcaldia_sel = dropdown_alcaldia.value
    actividades_sel = select_actividad.value
    temp_gdf = gdf.copy()

    if alcaldia_sel != 'Todas':
        temp_gdf = temp_gdf[temp_gdf['municipio'] == alcaldia_sel]
    if actividades_sel:
        pattern = '|'.join([re.escape(str(act).strip().lower()) for act in actividades_sel])
        temp_gdf = temp_gdf[temp_gdf['nombre_act'].str.lower().str.contains(pattern, na=False, regex=True) | temp_gdf['codigo_act'].astype(str).str.contains(pattern, na=False, regex=True)]

    current_gdf = temp_gdf
    visual_limit = 2000
    display_gdf = current_gdf.head(visual_limit)
    markers = []

    for idx, row in display_gdf.iterrows():
        html_content = f"""
        <div style="font-family:sans-serif; font-size:12px; color: #000000 !important; font-weight: normal; min-width:200px;">
            <b>{clean_val(row.get('nom_estab'))}</b><br>
            <i>{clean_val(row.get('nombre_act'))}</i><hr style="margin:4px 0;">
            <b>Personal:</b> {clean_val(row.get('per_ocu'))}<br>
            <b>Teléfono:</b> {clean_val(row.get('telefono'))}<br>
            <b>Correo:</b> {clean_val(row.get('correoelec'))}<br>
            <b>Web:</b> {clean_val(row.get('www'))}
        </div>
        """
        #icon = AwesomeIcon(name='building', marker_color=get_color(row['nombre_act']), icon_color='white')
        marker = Marker(location=(row['latitud'], row['longitud']))#, icon=icon)
        marker.popup = widgets.HTML(value=html_content)
        marker.on_click(create_marker_click_handler(row.to_dict()))
        markers.append(marker)

    marker_cluster.markers = tuple(markers)
    with out_mensajes:
        out_mensajes.clear_output()
        msg = f" Filtro aplicado. Registros encontrados: {len(current_gdf)}."
        if len(current_gdf) > visual_limit:
            msg += f" (Mostrando solo {visual_limit} en el mapa para mantener rendimiento)."
        print(msg)

btn_filtrar.on_click(actualizar_mapa)

def manejar_dibujo(self, action, geo_json):
    if action == 'created':
        with out_resultados_poligono:
            out_resultados_poligono.clear_output()
            print("Calculando intersección espacial...")
            try:
                poligono_dibujado = shape(geo_json['geometry'])
                intersectados = current_gdf[current_gdf.geometry.within(poligono_dibujado)]
                print(f" Negocios dentro del área seleccionada: {len(intersectados)}")
            except Exception as e:
                print(f"Error procesando la geometría: {e}")

draw_control.on_draw(manejar_dibujo)

pwd_groq = widgets.Password(description='Groq API:', placeholder='GROQ_API_KEY')
txt_material = widgets.Text(description='Consulta:', placeholder='¿Qué deseas cotizar o consultar sobre el negocio?')
btn_agente = widgets.Button(description='Analizar / Cotizar', button_style='success')
out_agente = widgets.Output()
html_selected = widgets.HTML(value="<b>Negocio seleccionado:</b> Ninguno (Haz clic en un marcador)")

ui_agente = widgets.VBox([
    widgets.HTML("<h3>Agente RAG + Scraper</h3>"),
    pwd_groq,
    html_selected,
    txt_material,
    btn_agente,
    out_agente
], layout=widgets.Layout(width='30%', padding='10px', border='1px solid #ccc', margin='0 0 0 10px'))

class GraphState(TypedDict):
    business_info: dict
    material: str
    groq_api_key: str
    search_results: str
    draft: str

def check_website(state: GraphState):
    www = str(state["business_info"].get("www", ""))
    if www.lower() not in ["no disponible", "nan", ""] and len(www) > 4:
        return "search"
    return "draft"

async def search_node(state: GraphState):
    info = state["business_info"]
    www = str(info.get("www", "")).strip()

    if not www.startswith("http"):
        www = "https://" + www

    try:
        browser_cfg = BrowserConfig(headless=True)
        run_cfg = CrawlerRunConfig(word_count_threshold=10, exclude_external_links=True)

        async with AsyncWebCrawler(config=browser_cfg) as crawler:
            result = await crawler.arun(url=www, config=run_cfg)
            if result.success:
                state["search_results"] = result.markdown
            else:
                state["search_results"] = f"Fallo al extraer el sitio web con Crawl4AI: {result.error_message}"
    except Exception as e:
        state["search_results"] = f"Error general en el scraper asíncrono: {e}"

    return state

async def analyze_node(state: GraphState):
    info = state["business_info"]
    www = str(info.get("www", ""))
    material = state["material"]
    search_res = state.get("search_results", "")

    safe_context = search_res[:30000]

    try:
        llm = ChatGroq(api_key=state["groq_api_key"], model="openai/gpt-oss-120b", temperature=0.1)

        prompt = f"""
        Eres un asistente experto en análisis de datos web de compras e investigación B2B.
        El usuario ha solicitado lo siguiente: "{material}"

        Has realizado un scrapeo de la página web del proveedor ({www}).
        CONTENIDO EXTRAÍDO (Markdown):
        {safe_context}

        Reglas y Restricciones:
        1. Analiza los resultados y responde a la petición del usuario utilizando ÚNICAMENTE la información extraída.
        2. NO redactes un correo. Genera un reporte analítico de tu hallazgo (insumos, servicios, filosofía, precios).
        3. Si la respuesta no está en el contenido o no ofrecen eso, indica claramente que NO se encontraron datos referentes a eso en su web pública.
        4. Sé conciso, directo y estructura tu respuesta con viñetas. No alucines información.
        """
        res = await llm.ainvoke([SystemMessage(content="Eres un asistente analítico, objetivo y estricto con el contexto."), HumanMessage(content=prompt)])
        state["draft"] = res.content
    except Exception as e:
        state["draft"] = f"Error en LLM Groq (Análisis RAG): {e}"
    return state

async def draft_node(state: GraphState):
    info = state["business_info"]
    material = state["material"]

    try:
        llm = ChatGroq(api_key=state["groq_api_key"], model="openai/gpt-oss-120b")
        prompt = f"""
        Eres un asistente de compras experto. Redacta un correo profesional solicitando una cotización.

        Material/Servicio a consultar: {material}

        Datos del proveedor:
        - Nombre: {info.get('nom_estab', 'N/D')}
        - Teléfono: {info.get('telefono', 'N/D')}
        - Correo: {info.get('correoelec', 'N/D')}

        Instrucciones:
        1. Escribe el ASUNTO del correo.
        2. Redacta el CUERPO del correo.
        3. Si no tienen correo electrónico provisto, indica explícitamente y crea un guion de 3 renglones para contactarlo vía telefónica.
        """
        res = await llm.ainvoke([SystemMessage(content="Eres un asistente útil y profesional."), HumanMessage(content=prompt)])
        state["draft"] = res.content
    except Exception as e:
        state["draft"] = f"Error en LLM Groq (Draft): {e}"
    return state

workflow = StateGraph(GraphState)
workflow.add_node("search", search_node)
workflow.add_node("analyze", analyze_node)
workflow.add_node("draft", draft_node)

workflow.set_conditional_entry_point(check_website, {"search": "search", "draft": "draft"})
workflow.add_edge("search", "analyze")
workflow.add_edge("analyze", END)
workflow.add_edge("draft", END)

app_graph = workflow.compile()

def ejecutar_agente(b):
    with out_agente:
        out_agente.clear_output()
        if not selected_business:
            print(" Selecciona un negocio en el mapa primero.")
            return
        if not txt_material.value:
            print(" Ingresa tu consulta o el material.")
            return
        if not pwd_groq.value:
            print(" Ingresa tu API Key de Groq.")
            return

        print(" Ejecutando IA (Extrayendo web y procesando)...")
        state = {
            "business_info": selected_business,
            "material": txt_material.value,
            "groq_api_key": pwd_groq.value,
            "search_results": "",
            "draft": ""
        }
        try:
            loop = asyncio.get_event_loop()
            result = loop.run_until_complete(app_graph.ainvoke(state))

            print("\n")
            print(" RESPUESTA DEL AGENTE RAG ")
            print("\n")
            print(result["draft"])
        except Exception as e:
            print(f" Error durante la ejecución del agente: {e}")

btn_agente.on_click(ejecutar_agente)

ui_filtros = widgets.HBox([dropdown_alcaldia, select_actividad, btn_filtrar])
mapa_container = widgets.VBox([
    ui_filtros,
    out_mensajes,
    m,
    out_resultados_poligono
], layout=widgets.Layout(width='70%'))

dashboard = widgets.VBox([
    widgets.HTML("<h2>Mapa DENUE potenciado con Agentes</h2>"),
    widgets.HBox([mapa_container, ui_agente])
])

display(dashboard)
#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
```
